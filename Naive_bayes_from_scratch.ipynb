```python
import numpy as np
class GussianNaiveBayes:
  def fit (self,x,y):
    n_samples, n_feature=x.shape
    self._classes=np.unique(y)
    n_classes=len(self._classes)
    self._mean=np.zeros((n_samples,n_feature),dtype=np.float64)
    self._variance=np.zeros((n_samples,n_feature),dtype=np.float64)
    self._prior=np.zeros(n_samples,dtype=np.float64)


    #calculating mean, variance and prior
    for i,c in enumerate(self._classes):
      x_for_class_c=x[y==c]
      self._mean[i,:]=x_for_class_c.mean(axis=0)
      self._variance[i,:]=x_for_class_c.var(axis=0)
      self._prior[i] = x_for_class_c.shape[0] / float(n_samples)


    #calculating liklihood
  def _likelihood(self, class_idx, x):
    mean = self._mean[class_idx]
    var = self._variance[class_idx]
    epsilon = 1e-9
    num = np.exp(-(x - mean)**2 / (2 * (var + epsilon)))
    denom = np.sqrt(2 * np.pi * (var + epsilon))
    return num / denom


 #classification
  def _classify_sample(self,x):
      posteriors=[]
      for i,c in enumerate(self._classes):
        prior=np.log(self._prior[i])
        class_liklihood=np.sum(np.log(self._likelihood(i,x)))
        posterior=prior+class_liklihood
        posteriors.append(posterior)
      return self._classes[np.argmax(posteriors)]
    #prediction
  def predict(self,x):
      y_pred=[self._classify_sample(i) for i in x]
      return np.array(y_pred)


```

```python
#load data
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
from sklearn.naive_bayes import GaussianNB
x,y=make_classification(n_samples=10000,n_features=20,n_classes=2,random_state=123)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=123)
```

```python
# del GussianNaiveBayes
```

```python
gnb=GussianNaiveBayes()

```

```python
import time
start_time=time.perf_counter()
gnb.fit(x_train,y_train)
y_pred=gnb.predict(x_test)
accuracy_score(y_test,y_pred)*100
end_time=time.perf_counter()
print(f'duration of manual model:{end_time-start_time}')
```

```python
sky_gnb=GaussianNB()
sky_gnb.fit(x_train,y_train)
y_pred=sky_gnb.predict(x_test)
accuracy_score(y_test,y_pred)*100
```
